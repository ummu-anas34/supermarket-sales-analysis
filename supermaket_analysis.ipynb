```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 500

df = pd.DataFrame({
    "Invoice_ID": [f"INV{i:04d}" for i in range(1, n + 1)],
    "Date": pd.to_datetime(
        np.random.choice(
            pd.date_range("2026-01-01", "2026-06-30"),
            n
        )
    ),
    "Branch": np.random.choice(["A", "B", "C"], n),
    "City": np.random.choice(
        ["Lagos", "Abuja", "Ibadan"], n
    ),
    "Gender": np.random.choice(
        ["Male", "Female"], n
    ),
    "Product_Line": np.random.choice(
        [
            "Food and Beverages",
            "Fashion Accessories",
            "Electronic Accessories",
            "Health and Beauty",
            "Home and Lifestyle",
            "Sports and Travel"
        ],
        n
    ),
    "Unit_Price": np.round(
        np.random.uniform(500, 20000, n), 2
    ),
    "Quantity": np.random.randint(1, 11, n),
    "Payment": np.random.choice(
        ["Cash", "Card", "Bank Transfer"], n
    )
})

df["Total_Sales"] = (
    df["Unit_Price"] * df["Quantity"]
).round(2)

df.to_csv("supermarket_sales.csv", index=False)

print("Dataset created successfully!")
print("Number of transactions:", len(df))

df.head(20)
```

```python
average_transaction = df["Total_Sales"].mean()
print("Average Transaction Value: ",
      round(average_transaction, 2))
```
